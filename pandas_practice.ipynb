## 1. Series и DataFrame

Таблица из десяти заказов: order_id, customer, city, category, price, quantity, order_date.

- создай таблицу сначала из словаря, затем из списка словарей
- изучи `shape`, `columns`, `index` и `dtypes`
- выведи первые и последние строки
- выбери один столбец как Series и несколько столбцов как DataFrame
- переименуй два столбца и измени порядок колонок
- добавь новый заказ и удали одну строку

```python
# блок 1
# 1.1 из словаря: ключ — название столбца, значение — список значений этого столбца
orders_dict = {
    "order_id":   [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "customer":   ["Анна", "Борис", "Вера", "Глеб", "Анна", "Дина", "Борис", "Елена", "Вера", "Глеб"],
    "city":       ["Москва", "Казань", "Москва", "Самара", "Москва", "Казань", "Казань", "Самара", "Москва", "Самара"],
    "category":   ["Электроника", "Книги", "Одежда", "Электроника", "Книги", "Одежда", "Электроника", "Книги", "Электроника", "Одежда"],
    "price":      [25000, 800, 3500, 12000, 650, 4200, 18000, 1200, 9000, 2800],
    "quantity":   [1, 3, 2, 1, 4, 1, 2, 2, 1, 3],
    "order_date": ["2026-09-05", "2026-09-07", "2026-09-12", "2026-09-15", "2026-09-20",
                   "2026-09-23", "2026-09-27", "2026-10-01", "2026-10-04", "2026-10-08"],
}
df_from_dict = pd.DataFrame(orders_dict)

# 1.2 из списка словарей: один словарь — одна строка (как транзакции в трекере расходов)
orders_list = [
    {"order_id": 101, "customer": "Анна",  "city": "Москва", "category": "Электроника", "price": 25000, "quantity": 1, "order_date": "2026-09-05"},
    {"order_id": 102, "customer": "Борис", "city": "Казань", "category": "Книги",       "price": 800,   "quantity": 3, "order_date": "2026-09-07"},
    {"order_id": 103, "customer": "Вера",  "city": "Москва", "category": "Одежда",      "price": 3500,  "quantity": 2, "order_date": "2026-09-12"},
    {"order_id": 104, "customer": "Глеб",  "city": "Самара", "category": "Электроника", "price": 12000, "quantity": 1, "order_date": "2026-09-15"},
    {"order_id": 105, "customer": "Анна",  "city": "Москва", "category": "Книги",       "price": 650,   "quantity": 4, "order_date": "2026-09-20"},
    {"order_id": 106, "customer": "Дина",  "city": "Казань", "category": "Одежда",      "price": 4200,  "quantity": 1, "order_date": "2026-09-23"},
    {"order_id": 107, "customer": "Борис", "city": "Казань", "category": "Электроника", "price": 18000, "quantity": 2, "order_date": "2026-09-27"},
    {"order_id": 108, "customer": "Елена", "city": "Самара", "category": "Книги",       "price": 1200,  "quantity": 2, "order_date": "2026-10-01"},
    {"order_id": 109, "customer": "Вера",  "city": "Москва", "category": "Электроника", "price": 9000,  "quantity": 1, "order_date": "2026-10-04"},
    {"order_id": 110, "customer": "Глеб",  "city": "Самара", "category": "Одежда",      "price": 2800,  "quantity": 3, "order_date": "2026-10-08"},
]
df = pd.DataFrame(orders_list)
print("таблицы одинаковые:", df.equals(df_from_dict))

# 1.3 устройство таблицы
print("\nshape:", df.shape)
print("columns:", df.columns.tolist())
print("index:", df.index)
print("\ndtypes:"); print(df.dtypes)

# 1.4 первые и последние строки
display(df.head(3))
display(df.tail(2))

# 1.5 один столбец → Series, несколько → DataFrame
city = df["city"]
two_cols = df[["customer", "price"]]
print(type(city).__name__, city.shape, "|", type(two_cols).__name__, two_cols.shape)

# 1.6 переименовать два столбца и поменять порядок колонок (на копии, df не трогаем)
demo = df.rename(columns={"customer": "client", "order_date": "date"})
demo = demo[["order_id", "date", "client", "city", "category", "price", "quantity"]]
print(demo.columns.tolist())

# 1.7 добавить заказ и удалить строку
new_order = pd.DataFrame([{"order_id": 111, "date": "2026-10-10", "client": "Жанна", "city": "Пермь",
                           "category": "Книги", "price": 950, "quantity": 2}])
demo = pd.concat([demo, new_order], ignore_index=True)
demo = demo[demo["order_id"] != 105]            # удалить заказ 105: оставляем всё, кроме него
print("строк в demo:", len(demo), "| заказ 111 есть:", 111 in demo["order_id"].values, "| 105 нет:", 105 not in demo["order_id"].values)
display(demo.tail(3))

# рабочая таблица для блоков 2–6: номер заказа — в индекс
orders = df.set_index("order_id")
display(orders.head(3))
```
