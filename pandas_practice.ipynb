## 2. Выбор данных и фильтрация


- выбери строки по индексам через `loc` и по позициям через `iloc`
- найди заказы дороже заданной суммы
- выбери заказы из нескольких городов через `isin`
- отфильтруй строки сразу по нескольким условиям
- найди товары, название категории которых содержит заданную подстроку
- отсортируй данные сначала по цене, затем по количеству
- выбери три заказа с максимальной итоговой стоимостью

```python
# блок 2
# 2.1 loc — по подписи (номер заказа), iloc — по позиции (какая по счёту)
display(orders.loc[104])                                  # один заказ → Series
display(orders.loc[[101, 105, 109], ["customer", "price"]])  # список строк × список столбцов
display(orders.loc[102:104])                              # по подписям: 104 ВКЛЮЧИТЕЛЬНО
display(orders.iloc[0:3])                                 # по позициям: 3 НЕ включается
display(orders.iloc[-1])                                  # последняя строка

# 2.2 заказы дороже заданной суммы
min_price = 5000
display(orders[orders["price"] > min_price])

# 2.3 заказы из нескольких городов
display(orders[orders["city"].isin(["Казань", "Самара"])])

# 2.4 несколько условий сразу: каждое в скобках, между ними & (и) или | (или)
display(orders[(orders["category"] == "Электроника") & (orders["price"] < 15000)])

# 2.5 категория содержит подстроку
display(orders[orders["category"].str.contains("ни", case=False)])

# 2.6 сортировка: сначала по цене (по убыванию), при равной цене — по количеству
display(orders.sort_values(["price", "quantity"], ascending=[False, True]))
display(orders.sort_values(["city", "price"], ascending=[True, False])[["city", "price"]])

# 2.7 три заказа с максимальной итоговой стоимостью (price × quantity)
top3 = orders.assign(total=orders["price"] * orders["quantity"]).nlargest(3, "total")
display(top3[["customer", "price", "quantity", "total"]])
```

### Вывод по блоку 2
loc выбирает по подписи, iloc — по позиции. Срез loc включает правую границу, iloc — нет.
Дороже 5 000 только заказы электроники: это самая дорогая категория.
Сортировка по нескольким столбцам работает как «по первому, при равенстве — по второму». У нас цены уникальны, поэтому второй ключ не понадобился.
Самый дорогой товар и самый дорогой заказ — разные вещи. Самая высокая цена у заказа 101 (25 000), но по итоговой сумме лидирует 107: 18 000 × 2 = 36 000. Для выручки важен price × quantity, а не цена.

## 3. Очистка данных

Намеренно добавляем пропуски, дубликаты, отрицательную цену и даты в неправильном формате.

- найди количество пропусков в каждом столбце
- выбери подходящую стратегию заполнения или удаления пропусков
- найди и удали полные и частичные дубликаты
- преобразуй числовые столбцы и даты в правильные типы
- обнаружь некорректные значения и исправь либо исключи их
- сравни размер и статистики таблицы до и после очистки
- объясни, почему очистку важно выполнять до анализа и обучения модели

```python
# блок 3
# 3.1 «портим» данные, как будто они пришли из CSV-выгрузки
dirty = df.astype({"price": "str", "quantity": "float"})   # цена строкой, количество дробным (чтобы влез NaN)
dirty.loc[0, "customer"] = "  анна "          # лишние пробелы и регистр (поправим в блоке 4)
dirty.loc[1, "order_date"] = "07.09.2026"     # другой формат даты
dirty.loc[2, "quantity"] = np.nan             # пропуск
dirty.loc[3, "price"] = "12 000"              # пробел внутри числа
dirty.loc[4, "city"] = "москва "              # регистр и пробел (поправим в блоке 4)
dirty.loc[5, "price"] = "-4200"               # отрицательная цена
dirty.loc[7, "order_date"] = "2026/10/01"     # ещё один формат даты
dirty.loc[8, "customer"] = np.nan             # пропуск
dirty.loc[9, "order_date"] = "вчера"          # мусор вместо даты
full_dup = dirty.loc[[1]]                                   # точная копия заказа 102
partial_dup = dirty.loc[[6]].assign(quantity=3.0)           # заказ 107 пришёл ещё раз, с исправленным количеством
dirty = pd.concat([dirty, full_dup, partial_dup], ignore_index=True)
before = dirty.copy()
display(dirty)

# 3.2 сколько пропусков в каждом столбце
print("пропуски:\n", dirty.isna().sum(), sep="")

# 3.3 дубликаты: полные и частичные (один order_id — разные данные)
print("\nполных дублей:", dirty.duplicated().sum())
dirty = dirty.drop_duplicates()
print("частичных дублей по order_id:", dirty.duplicated(subset="order_id").sum())
display(dirty[dirty["order_id"].duplicated(keep=False)][["order_id", "customer", "quantity"]])
dirty = dirty.drop_duplicates(subset="order_id", keep="last")   # последняя выгрузка — актуальная

# 3.4 типы: цена → число, дата → datetime
dirty["price"] = pd.to_numeric(dirty["price"].str.replace(" ", ""), errors="coerce")
raw = dirty["order_date"]
dirty["order_date"] = (pd.to_datetime(raw, format="%Y-%m-%d", errors="coerce")
                       .fillna(pd.to_datetime(raw, format="%d.%m.%Y", errors="coerce"))
                       .fillna(pd.to_datetime(raw, format="%Y/%m/%d", errors="coerce")))
print("типы после:\n", dirty.dtypes, sep="")
print("не распознаны даты:", raw[dirty["order_date"].isna()].tolist())

# 3.5 некорректные значения и пропуски: решение по каждому
display(dirty[dirty["price"] <= 0][["order_id", "price"]])
dirty = dirty[dirty["price"] > 0]                       # цену не угадать — исключаем
dirty = dirty.dropna(subset=["order_date"])             # заказ без даты не положить на время — исключаем
dirty["quantity"] = dirty["quantity"].fillna(dirty["quantity"].median()).astype(int)   # медиана количества
dirty["customer"] = dirty["customer"].fillna("Неизвестно")   # выручка реальная, клиента не знаем
clean = dirty.sort_values("order_id").reset_index(drop=True)

# 3.6 до и после
print("строк: до", len(before), "→ после", len(clean))
print("пропусков: до", before.isna().sum().sum(), "→ после", clean.isna().sum().sum())
raw_price = pd.to_numeric(before["price"].str.replace(" ", ""), errors="coerce")
print("цена до: среднее", round(raw_price.mean()), "| мин", raw_price.min())
print("цена после: среднее", round(clean["price"].mean()), "| мин", clean["price"].min())
display(clean)
```

В 12 строках нашлось 6 видов проблем, и isna() показал только часть: битые даты и отрицательная цена видны лишь после перевода типов и проверки значений.
На каждую проблему своё решение: дубли удаляем, неустранимые ошибки (цена, дата) исключаем, восстановимые пропуски заполняем медианой или пометкой «Неизвестно».
После очистки осталось 8 строк из 12. Средняя цена выросла с 7 296 до 8 769, потому что ушла отрицательная цена и задвоенные заказы.
Чистить нужно до анализа и обучения: иначе дубли завысят выручку, отрицательная цена исказит среднее, а модель выучит ошибки ввода как закономерности.

## 4. Новые признаки и преобразования


- создай столбец `total_price = price × quantity`
- выдели из даты год, месяц, день недели и признак выходного дня
- разбей стоимость заказа на категории low, medium и high через `pd.cut`
- приведи строковые значения к единому регистру и убери лишние пробелы
- создай признак длины имени клиента
- используй `map`, `replace` и `assign` минимум по одному разу
- сравни обычный `apply` с векторизованной операцией

```python
# блок 4
import time

fe = clean.copy()

# 4.1 строки: убрать лишние пробелы и привести к единому регистру
fe["customer"] = fe["customer"].str.strip().str.title()
fe["city"] = fe["city"].str.strip().str.title()

# 4.2 новые признаки через assign: итог, части даты, выходной, длина имени
fe = fe.assign(
    total_price=fe["price"] * fe["quantity"],
    year=fe["order_date"].dt.year,
    month=fe["order_date"].dt.month,
    weekday=fe["order_date"].dt.day_name(),
        is_weekend=fe["order_date"].dt.day_of_week >= 5,
    name_len=fe["customer"].str.len(),
)

# 4.3 категории стоимости через pd.cut
fe["size"] = pd.cut(fe["total_price"], bins=[0, 3000, 15000, np.inf], labels=["low", "medium", "high"])

# 4.4 map и replace: похожи, но ведут себя по-разному
short = {"Москва": "МСК", "Казань": "КЗН"}                 # Самары в словаре нет
fe["city_map"] = fe["city"].map(short)                       # нет в словаре → NaN
fe["city_replace"] = fe["city"].replace(short)               # нет в словаре → остаётся как было
fe["weekday_ru"] = fe["weekday"].map({"Monday": "Пн", "Tuesday": "Вт", "Wednesday": "Ср", "Thursday": "Чт",
                                      "Friday": "Пт", "Saturday": "Сб", "Sunday": "Вс"})
display(fe[["order_id", "customer", "city", "total_price", "year", "month", "weekday_ru", "is_weekend", "size", "name_len"]])
display(fe[["city", "city_map", "city_replace"]].drop_duplicates())

# 4.5 apply против векторной операции
same = fe.apply(lambda row: row["price"] * row["quantity"], axis=1).equals(fe["total_price"])
print("apply по строкам даёт тот же total_price:", same)

big = pd.Series(np.random.default_rng(0).integers(100, 30000, 1_000_000))
t0 = time.perf_counter(); r1 = big.apply(lambda x: x * 1.2); t_apply = time.perf_counter() - t0
t0 = time.perf_counter(); r2 = big * 1.2;                   t_vec = time.perf_counter() - t0
print(f"1 000 000 значений: apply {t_apply:.3f} с, векторно {t_vec:.4f} с → в {t_apply / t_vec:.0f} раз быстрее; "
      f"результаты равны: {np.allclose(r1, r2)}")
```

### Вывод по блоку 4

Из даты получились признаки, которые модель может использовать: месяц, день недели, выходной. 5 из 8 заказов сделаны в выходные — это может пригодиться для прогноза нагрузки.
pd.cut превращает число в категорию. Но границы мы выбрали сами, и от них зависит, сколько заказов попадёт в «high».
map подходит, когда словарь покрывает все значения: пропуск сразу покажет, что чего-то не хватает. replace — когда нужно поменять только часть значений.
Подвох с заглушкой: у «Неизвестно» name_len = 10 — это длина слова-заглушки, а не имени. Заполненные пропуски могут портить производные признаки, поэтому такие строки в ML лучше помечать отдельным признаком.
Векторная операция в ~50 раз быстрее apply. apply оставляем для того, что векторно не записать.

## 5. GroupBy и агрегации


- посчитай сумму и средний чек по городам
- найди количество заказов и уникальных клиентов по категориям
- рассчитай сразу несколько статистик через `agg`
- найди самый дорогой заказ внутри каждой категории
- добавь к каждой строке среднюю стоимость заказа её категории через `transform`
- вычисли долю каждого заказа в общей выручке своего города
- отсортируй группы по выручке и сформулируй минимум три вывода

```python
# блок 5
pd.set_option("display.precision", 2)

# 5.1 выручка и средний чек по городам (сразу отсортировано по выручке)
by_city = (fe.groupby("city")["total_price"]
             .agg(revenue="sum", avg_check="mean", orders="count")
             .sort_values("revenue", ascending=False))
display(by_city)

# 5.2 количество заказов и уникальных клиентов по категориям
by_cat = fe.groupby("category").agg(orders=("order_id", "count"), customers=("customer", "nunique"))
display(by_cat)

# 5.3 несколько статистик сразу
display(fe.groupby("category")["total_price"].agg(["count", "sum", "mean", "median", "min", "max"]))

# 5.4 самый дорогой заказ внутри каждой категории
top_idx = fe.groupby("category")["total_price"].idxmax()        # номер строки с максимумом в каждой группе
display(fe.loc[top_idx, ["category", "order_id", "customer", "total_price"]])

# 5.5 средняя стоимость заказа его категории — в каждую строку
fe["cat_avg"] = fe.groupby("category")["total_price"].transform("mean")
fe["vs_cat_avg"] = fe["total_price"] / fe["cat_avg"]            # во сколько раз дороже среднего по категории

# 5.6 доля заказа в выручке своего города
fe["city_share"] = fe["total_price"] / fe.groupby("city")["total_price"].transform("sum")
display(fe[["order_id", "city", "category", "total_price", "cat_avg", "vs_cat_avg", "city_share"]])
print("доли внутри каждого города в сумме дают 1:", fe.groupby("city")["city_share"].sum().round(6).eq(1).all())
```
